# Marketplace health

## tl;dr
Track GMV proxy, orders, active sellers and AOV at monthly grain.


## Context & Methods

GMV is a proxy from order-item price. Orders are distinct order IDs and active sellers are sellers with order-item activity.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
root = Path.cwd().parent
monthly = pd.read_csv(root / 'reports/tables/mart_marketplace_monthly.csv', parse_dates=['purchase_month'])
monthly.tail()
```

## Results

### Trend


```python
display(monthly[['purchase_month','gmv_proxy','orders','active_sellers','aov']])
fig, ax = plt.subplots(figsize=(10,4))
ax.plot(monthly['purchase_month'], monthly['gmv_proxy'], label='GMV proxy')
ax.set_ylabel('GMV proxy')
ax2 = ax.twinx(); ax2.plot(monthly['purchase_month'], monthly['orders'], color='tab:orange', label='Orders')
ax2.set_ylabel('orders')
ax.set_title('Monthly marketplace health')
fig.autofmt_xdate(); plt.show()
```

## Takeaways

Use the decomposition table beside the trend to distinguish volume, active-seller and AOV movements; do not call the decomposition causal.
