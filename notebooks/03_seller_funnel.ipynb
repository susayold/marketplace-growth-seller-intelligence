# Seller acquisition and activation

## tl;dr
Evaluate acquisition origins with both conversion and downstream seller value.


## Context & Methods

Funnel linkage is partial. A converted lead is not automatically a selling seller; downstream metrics use matched seller IDs.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
root = Path.cwd().parent
channel = pd.read_csv(root / 'reports/tables/mart_acquisition_channel.csv')
activation = pd.read_csv(root / 'reports/tables/activation_summary.csv')
channel
```

## Results

### Conversion versus downstream value


```python
display(channel)
fig, ax = plt.subplots(figsize=(9,4))
ax.scatter(channel['conversion_rate']*100, channel['gmv_per_matched_seller'])
for _, r in channel.iterrows(): ax.annotate(str(r['origin_group']), (r['conversion_rate']*100, r['gmv_per_matched_seller']))
ax.set_xlabel('Lead conversion rate (%)'); ax.set_ylabel('GMV proxy per matched seller'); ax.set_title('Channel quality: volume, conversion and value')
plt.show()
display(activation)
```

## Takeaways

Channel decisions should include denominator, conversion, seller match rate and downstream value. Activation windows use the valid seller-link population.
