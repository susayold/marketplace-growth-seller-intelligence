# Operations and customer experience

## tl;dr
Late delivery is strongly associated with lower reviews in this historical sample, but the result is not causal.


## Context & Methods

The table is at order level. Seller-level blame is avoided because orders may contain multiple sellers.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
root = Path.cwd().parent
experience = pd.read_csv(root / 'reports/tables/mart_order_experience.csv')
experience
```

## Results

### Delivery bucket comparison


```python
display(experience)
fig, axes = plt.subplots(1,2,figsize=(10,4))
labels = ['on time/early','late']
axes[0].bar(labels, experience['avg_review_score']); axes[0].set_ylabel('average review score')
axes[1].bar(labels, experience['low_review_rate']*100); axes[1].set_ylabel('low review rate (%)')
fig.suptitle('Customer experience by delivery outcome'); plt.show()
```

## Takeaways

Prioritize high-volume late-delivery segments for investigation and experiments; keep the claim associative.
